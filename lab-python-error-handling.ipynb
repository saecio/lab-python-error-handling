````python
products = ["t-shirt", "mug", "hat", "book", "keychain"]
````

````python
def initialize_inventory(products):
    """Create the inventory, validating each quantity entered by the user."""
    inventory = {}

    for product in products:
        quantity_is_valid = False

        while not quantity_is_valid:
            try:
                quantity = int(input(f"Enter the quantity of {product}s available: "))

                if quantity < 0:
                    raise ValueError("Quantity cannot be negative.")

                inventory[product] = quantity
                quantity_is_valid = True

            except ValueError as error:
                print(f"Invalid quantity: {error} Please try again.")

    return inventory
````

2. Modify the `calculate_total_price` function to include error handling.
   - If the user enters an invalid price (e.g., a negative value or a non-numeric value), display an error message and ask them to re-enter the price for that product.
   - Use a try-except block to handle the error and continue prompting the user until a valid price is entered.

Note: as per exercise, this should be executed prior to the next point. Per logic, it is easier to do afterwards so the parameter will be taken from next exercise.

````python
def calculate_total_price(customer_orders):
    """Ask for each product price and return the total order price."""
    total_price = []

    for item in customer_orders:
        price_is_valid = False

        while not price_is_valid:
            try:
                price = float(input(f"Please add a price for {item}: "))

                if price < 0:
                    raise ValueError("Price cannot be negative.")

                total_price.append(price)
                print(f"Price for {item} has been added.")
                price_is_valid = True

            except ValueError as error:
                print(f"Invalid price: {error} Please try again.")

    return sum(total_price)
````

3. Modify the `get_customer_orders` function to include error handling.
   - If the user enters an invalid number of orders (e.g., a negative value or a non-numeric value), display an error message and ask them to re-enter the number of orders.
   - If the user enters an invalid product name (e.g., a product name that is not in the inventory), or that doesn't have stock available, display an error message and ask them to re-enter the product name. *Hint: you will need to pass inventory as a parameter*
   - Use a try-except block to handle the error and continue prompting the user until a valid product name is entered.


````python
def get_customer_orders(inventory):
    customer_orders = set()

    # Ask for the number of products
    while True:
        try:
            num_orders = int(input("How many products would you like to order?: "))

            if num_orders <= 0:
                print("The number of orders must be greater than 0.")
                continue

            break

        except ValueError:
            print("Invalid input. Please enter a whole number.")

    # Ask for each product
    for _ in range(num_orders):

        while True:
            try:
                order = input("Please enter a product name: ").strip().lower()

                if order not in inventory:
                    raise ValueError("That product is not in the inventory.")

                if inventory[order] <= 0:
                    raise ValueError("That product is out of stock.")

                customer_orders.add(order)
                print(f"'{order}' has been added to your order!")
                break

            except ValueError as error:
                print(f"Invalid product: {error}")
                print("Please try again.")

    return customer_orders
````

````python
customer_orders = get_customer_orders(inventory)
````
